```python
def naive(P, Q):
    """
    A naive solution for finding pairwise distances between points in P and Q

    Args:
        P: numpy array of shape=(p, 2)
        Q: numpy array of shape=(q, 2)
    Returns:
        D: numpy array of shape=(p, q)

    >>> naive(np.array([[0, 1]]), np.array([[2, 3], [4, 5]]))
    array([[2.82842712, 5.65685425]])
    """
    # ***************************************************
    # créer une variable qui store les pairwise distances, dans un dictionnaire ? NON => dans un tableau
    # faire deux boucles imbriquées : itère sur P puis sur Q
    # test si c'est équidistance q et p, si oui (if condition satisfaite) alors ajoute à cette variable
    # puis return la variable 
    D = np.zeros((P.shape[0], Q.shape[0])) # tableau à retourner de taille (pxq)
    for i, point_i in enumerate(P):
        for j, point_j in enumerate(Q):
            dist = np.sqrt(np.sum(pow((point_i - point_j),2)))
            D[i, j] = dist
            
    return D


test(naive)
```

### Use matching indices

Instead of iterating through indices, one can use them directly to parallelize the operations with Numpy.

```python
rows, cols = np.indices((P.shape[0], Q.shape[0]))
print(rows, end="\n\n")
print(cols)
```

```python
print(P[rows.ravel()], end="\n\n")
print(Q[cols.ravel()])
```

```python
def with_indices(P, Q):
    """
    An optimized solution using matching indices

    Args:
        P: numpy array of shape=(p, 2)
        Q: numpy array of shape=(q, 2)
    Returns:
        D: numpy array of shape=(p, q)

    >>> with_indices(np.array([[0, 1]]), np.array([[2, 3], [4, 5]]))
    array([[2.82842712, 5.65685425]])
    """
    
    rows, cols = np.indices((P.shape[0], Q.shape[0]))
    diff = P[rows] - Q[cols]
    D = np.sqrt(np.sum(diff**2, axis=2))
    return D

test(with_indices)
```

### Use a library

`scipy` is the equivalent of matlab toolboxes and have a lot to offer. Actually the pairwise computation is part of the library through the `spatial` module.

```python
from scipy.spatial.distance import cdist


def scipy_version(P, Q):
    """
    A solution using scipy

    Args:
        P: numpy array of shape=(p, 2)
        Q: numpy array of shape=(q, 2)

    Returns:
        D: numpy array of shape=(p, q)

    >>> scipy_version(np.array([[0, 1]]), np.array([[2, 3], [4, 5]]))
    array([[2.82842712, 5.65685425]])
    """
    return cdist(P, Q)
```

### Numpy Magic

```python
def tensor_broadcasting(P, Q):
    """
    A solution using tensor broadcasting

    Args:
        P: numpy array of shape=(p, 2)
        Q: numpy array of shape=(q, 2)

    Returns:
        D: numpy array of shape=(p, q)

    >>> tensor_broadcasting(np.array([[0, 1]]), np.array([[2, 3], [4, 5]]))
    array([[2.82842712, 5.65685425]])
    """
    return np.sqrt(np.sum((P[:, np.newaxis, :] - Q[np.newaxis, :, :]) ** 2, axis=2))
```
